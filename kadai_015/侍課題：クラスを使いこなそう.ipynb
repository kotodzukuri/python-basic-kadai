```python
# Humanクラスを定義
class Human:
    def __init__(self, name, age):
        self.name = name  # 名前の属性
        self.age = age    # 年齢の属性

    def printinfo(self):
        # 属性を出力するメソッド
        print(f"名前：{self.name}, 年齢：{self.age}")

# インスタンスを作成して変数に代入
human1 = Human("さやか", 52)

# メソッドを呼び出して情報を出力
human1.printinfo()
```
