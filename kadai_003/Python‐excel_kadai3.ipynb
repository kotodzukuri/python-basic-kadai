```python
import pandas as pd
from openpyxl import load_workbook
from openpyxl.styles import PatternFill
from io import StringIO

# ----------------------------------------------------
# 1. データフレームの作成と連結
# ----------------------------------------------------

# 2022年と2023年の売上データをCSV形式の文字列として定義
# 実際には提供されたファイルの内容を読み込むことを想定していますが、
# 今回は添付された内容を直接コード内に含めます。
# ファイルから読み込む場合は pd.read_excel('ファイル名.xlsx', sheet_name='Sheet1') などを使用してください。

# kadai_012.zip/課題3/2022_年間売上表.xlsx - Sheet1.csv の内容
csv_data_2022 = """売上年,商品,金額（千円）
2022,紅茶,1200
2022,ハチミツ,1000
2022,セッケン,1300
2022,紅茶,800
2022,アロマキャンドル,900
2022,ハチミツ,300
2022,ハチミツ,560
2022,ワイングラス,850
2022,アロマキャンドル,810
2022,包丁,940
2022,水筒,1150
"""
# [cite: 1]
df_2022 = pd.read_csv(StringIO(csv_data_2022))

# kadai_012.zip/課題3/2023_年間売上表.xlsx - Sheet1.csv の内容
csv_data_2023 = """売上年,商品,金額（千円）
2023,紅茶,1200
2023,ハチミツ,1000
2023,水筒,250
2023,コーヒー豆,860
2023,セッケン,1300
2023,紅茶,670
2023,アロマキャンドル,900
2023,ワイングラス,850
2023,包丁,940
2023,フライパン,670
2023,水筒,1150
"""
# [cite: 2]
df_2023 = pd.read_csv(StringIO(csv_data_2023))

# データを連結
df_combined = pd.concat([df_2022, df_2023], ignore_index=True)

# ----------------------------------------------------
# 2. データのグループ化と集約
# ----------------------------------------------------

# 「商品」と「売上年」ごとにグループ化し、「金額（千円）」を合計
df_aggregated = df_combined.groupby(['商品', '売上年'], as_index=False).agg(
    {'金額（千円）': 'sum'}
)

# 見やすいように「商品」をキーにしてソート
df_aggregated = df_aggregated.sort_values(by=['商品', '売上年']).reset_index(drop=True)

# ----------------------------------------------------
# 3. Excelファイルへの書き込み (pandas)
# ----------------------------------------------------

output_file_name = '売上集計表.xlsx'

# インデックスなしで新規Excelファイルに書き込み
df_aggregated.to_excel(output_file_name, index=False, sheet_name='集計結果')

# ----------------------------------------------------
# 4. 見た目の調整 (openpyxl)
# ----------------------------------------------------

# openpyxlで書き込んだExcelファイルを読み込み
wb = load_workbook(output_file_name)
ws = wb['集計結果']

# ヘッダー部分（1行目）を薄いグレー（#F2F2F2）に設定
header_fill = PatternFill(start_color='F2F2F2', end_color='F2F2F2', fill_type='solid')

# 1行目の各セルに色を適用
for cell in ws[1]:
    cell.fill = header_fill

# 列幅の調整（任意）
# 'A'列（商品）は広めに、'B'列（売上年）、'C'列（金額）は標準的
ws.column_dimensions['A'].width = 15
ws.column_dimensions['B'].width = 10
ws.column_dimensions['C'].width = 15

# 変更を保存
wb.save(output_file_name)

print(f"処理が完了しました。ファイル '{output_file_name}' が作成されました。")
```
